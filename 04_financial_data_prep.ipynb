##  Fetch filing history(Company House) 

```python
API_KEY = "c139ab29-2252-47a0-99fb-ee79c316ba53"
COMPANY_NUMBER = "12743269"  # Revolut Group Holdings Ltd

url = f"https://api.company-information.service.gov.uk/company/{COMPANY_NUMBER}/filing-history"
response = requests.get(url, auth=(API_KEY, ""))

if response.status_code == 200:
    data = response.json()
    filings = pd.json_normalize(data["items"])
    print(f"Fetched {len(filings)} filings")
    filings.head()
else:
    print(f"Error: {response.status_code}")
    print(response.text)
```

```python
accounts_filings = filings[filings["type"] == "AA"][["date", "description", "links.document_metadata"]]
accounts_filings
```

## Fetching Revolut's Annual Report from Companies House

**Goal:** Download Revolut Group Holdings Ltd's official annual accounts (primary source, not press coverage).

**Process:**
1. `GET /company/{number}/filing-history?category=accounts` → list of accounts filings (dates + metadata links)
2. `GET {metadata_link}` → info *about* the document (pages, date, and a `document` URL — not the file itself)
3. `GET {document_url}` with header `Accept: application/pdf` → the actual PDF bytes
4. Save bytes to `.pdf` with `open(path, "wb")`

**Auth:** API key goes in `auth=(API_KEY, "")` — key as username, empty password.

**Result:** `revolut_accounts_2025.pdf` (~11.4 MB, 207 pages) — Revolut's real FY2025 annual report, filed with UK Companies House (company no. 12743269).
